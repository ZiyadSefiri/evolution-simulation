```python
from mutation import Truncate, Mutate , EvolutionStrategy , Experiment ,Individuel,Generation
```

```python
mutate = Mutate(0.05) 
truncate = Truncate( 0.5 )
evolution = EvolutionStrategy(truncate , mutate)
```

```python
from random import random
import numpy as np

def random_individual(p,N) :
    genome = np.array([1 if random()>p else 0 for i in range(N)])
    return Individuel(genome)
```

```python
random_individual(0.3 , 1000)
```

```python
def random_generation(N , n , p) :
    return Generation([random_individual(p,n) for _ in range(N)] )
```

```python
generation_0 = random_generation(500 , 50 , 0.2) 
experiment = Experiment(evolution_strategy= evolution , starting_generation=generation_0 , number_of_iterations=10)
```

```python
experiment.run_experiment()
```

```python
import matplotlib.pyplot as plt
```

```python
fig , ax = plt.subplots()
experiment.plot_fitness(fig ,ax)
```
